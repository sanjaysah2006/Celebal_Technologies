```python
from pyspark.sql.functions import current_timestamp, lit
```

```python
customers = spark.read \
.option("header", True) \
.option("inferSchema", True) \
.csv("/Volumes/workspace/default/fmcg_data/customers_dirty.csv")

customers.printSchema()

display(customers.limit(5))
```

```python
bronze_customers = customers \
.withColumn("source_file", lit("customers_dirty.csv")) \
.withColumn("ingestion_time", current_timestamp())
```

```python
bronze_customers.write \
.format("delta") \
.mode("overwrite") \
.save("/Volumes/workspace/default/fmcg_data/bronze/customers")
```

```python
display(
    spark.read.format("delta")
    .load("/Volumes/workspace/default/fmcg_data/bronze/customers")
)
```
