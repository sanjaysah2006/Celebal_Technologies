```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
bronze_customers = spark.read.format("delta").load(
    "/Volumes/workspace/default/fmcg_data/bronze/customers"
)

display(bronze_customers.limit(10))
```

```python
silver_customers = bronze_customers.dropDuplicates(["customer_id"])
```

```python
silver_customers = silver_customers.withColumn(
    "city",
    initcap(lower(col("city")))
)
```

```python
silver_customers = silver_customers.fillna({
    "email":"unknown@company.com"
})
```

```python
silver_customers.write \
.format("delta") \
.mode("overwrite") \
.save("/Volumes/workspace/default/fmcg_data/silver/customers")
```

```python
bronze_sales = spark.read.format("delta").load(
"/Volumes/workspace/default/fmcg_data/bronze/sales")
```

```python
silver_sales = bronze_sales.dropDuplicates(["sale_id"])
```

```python
silver_sales = silver_sales.filter(
    col("quantity") > 0
)
```

```python
silver_sales = silver_sales.fillna({
    "amount":0
})
```

```python
silver_sales = silver_sales.withColumn(
    "sale_date",
    to_date(col("sale_date"))
)
```

```python
silver_sales.write \
.format("delta") \
.mode("overwrite") \
.save("/Volumes/workspace/default/fmcg_data/silver/sales")
```

```python
products = spark.read.format("delta").load(
"/Volumes/workspace/default/fmcg_data/bronze/products")

products.write.format("delta").mode("overwrite").save(
"/Volumes/workspace/default/fmcg_data/silver/products")
```

```python
bronze_stores = spark.read.format("delta").load(
    "/Volumes/workspace/default/fmcg_data/bronze/stores"
)

silver_stores = bronze_stores.dropDuplicates(["store_id"])

silver_stores.write \
.format("delta") \
.mode("overwrite") \
.save("/Volumes/workspace/default/fmcg_data/silver/stores")

print("✅ Silver Stores Created")
```

```python
bronze_inventory = spark.read.format("delta").load(
    "/Volumes/workspace/default/fmcg_data/bronze/inventory"
)

silver_inventory = bronze_inventory.dropDuplicates(["product_id"])

silver_inventory.write \
.format("delta") \
.mode("overwrite") \
.save("/Volumes/workspace/default/fmcg_data/silver/inventory")

print("✅ Silver Inventory Created")
```

```python
bronze_promotions = spark.read.format("delta").load(
    "/Volumes/workspace/default/fmcg_data/bronze/promotions"
)

silver_promotions = bronze_promotions.dropDuplicates(["promotion_id"])

silver_promotions.write \
.format("delta") \
.mode("overwrite") \
.save("/Volumes/workspace/default/fmcg_data/silver/promotions")

print("✅ Silver Promotions Created")
```
