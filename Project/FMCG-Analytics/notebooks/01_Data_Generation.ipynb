```python
from faker import Faker
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

fake = Faker("en_IN")

random.seed(42)
np.random.seed(42)
```

```python
cities = [
    ("Delhi", "Delhi"),
    ("Mumbai", "Maharashtra"),
    ("Bangalore", "Karnataka"),
    ("Chennai", "Tamil Nadu"),
    ("Hyderabad", "Telangana"),
    ("Kolkata", "West Bengal"),
    ("Pune", "Maharashtra"),
    ("Lucknow", "Uttar Pradesh")
]

customers = []

for i in range(1,1001):

    city,state=random.choice(cities)

    customers.append({
        "customer_id":i,
        "customer_name":fake.name(),
        "city":city,
        "state":state,
        "phone":fake.phone_number(),
        "email":fake.email()
    })

customers_df=pd.DataFrame(customers)

customers_df.to_csv(
"/Volumes/workspace/default/fmcg_data/customers.csv",
index=False
)

customers_df.head()
```

```python
categories = [
    "Beverages",
    "Snacks",
    "Personal Care",
    "Household",
    "Dairy"
]

brands = [
    "Nestle",
    "Dabur",
    "HUL",
    "ITC",
    "Parle",
    "Amul"
]

products=[]

for i in range(1,501):

    category=random.choice(categories)

    products.append({
        "product_id":i,
        "product_name":fake.word().capitalize(),
        "category":category,
        "brand":random.choice(brands),
        "price":round(random.uniform(20,500),2)
    })

products_df=pd.DataFrame(products)

products_df.to_csv(
"/Volumes/workspace/default/fmcg_data/products.csv",
index=False
)

products_df.head()
```

```python
stores=[]

for i in range(1,101):

    city,state=random.choice(cities)

    stores.append({
        "store_id":i,
        "store_name":"Store_"+str(i),
        "city":city,
        "state":state
    })

stores_df=pd.DataFrame(stores)

stores_df.to_csv(
"/Volumes/workspace/default/fmcg_data/stores.csv",
index=False
)

stores_df.head()
```

```python
sales=[]

for i in range(1,10001):

    quantity=random.randint(1,10)

    price=random.uniform(20,500)

    sales.append({

        "sale_id":i,

        "customer_id":random.randint(1,1000),

        "product_id":random.randint(1,500),

        "store_id":random.randint(1,100),

        "quantity":quantity,

        "amount":round(quantity*price,2),

        "sale_date":fake.date_between(
            start_date="-2y",
            end_date="today"
        )

    })

sales_df=pd.DataFrame(sales)

sales_df.to_csv(
"/Volumes/workspace/default/fmcg_data/sales.csv",
index=False
)

sales_df.head()
```

```python
inventory=[]

for product in range(1,501):

    inventory.append({
        "product_id":product,
        "stock":random.randint(20,500)
    })

inventory_df=pd.DataFrame(inventory)

inventory_df.to_csv(
"/Volumes/workspace/default/fmcg_data/inventory.csv",
index=False
)

inventory_df.head()
```

```python
promotions=[]

for i in range(1,101):

    start=fake.date_between("-1y","today")
    end=start+timedelta(days=30)

    promotions.append({

        "promotion_id":i,

        "product_id":random.randint(1,500),

        "discount":random.choice([5,10,15,20,25,30]),

        "start_date":start,

        "end_date":end

    })

promotions_df=pd.DataFrame(promotions)

promotions_df.to_csv(
"/Volumes/workspace/default/fmcg_data/promotions.csv",
index=False
)

promotions_df.head()
```

```python
# Introduce data quality issues

# Duplicate customers
customers_dirty = pd.concat([customers_df, customers_df.iloc[:20]], ignore_index=True)

# Missing emails
customers_dirty.loc[
    customers_dirty.sample(25, random_state=42).index,
    "email"
] = None

# Inconsistent city names
customers_dirty["city"] = customers_dirty["city"].replace({
    "Delhi":"delhi",
    "Mumbai":"MUMBAI",
    "Bangalore":"bangalore"
})

customers_dirty.to_csv(
"/Volumes/workspace/default/fmcg_data/customers_dirty.csv",
index=False
)

print(customers_dirty.shape)
```
