## Spark vs MapReduce

MapReduce processes data by repeatedly reading and writing intermediate data to disk, which increases execution time.

Apache Spark performs most computations in memory, reducing disk I/O and making processing significantly faster.

Advantages of Spark:
- In-memory processing
- Faster execution
- Easy DataFrame API
- Supports iterative machine learning workloads

```python
from pyspark.sql import SparkSession


spark = SparkSession.builder \
    .appName("spark_basics") \
    .getOrCreate()
```

Step 2: Load CSV File

```python

df = spark.read.csv(
    "../data/dataset.csv",
    header=True,
    inferSchema=True
)


df.show(5)
```

```python

df.printSchema()
```

```python

print(df.columns)
```

Step 3: Data Cleaning

```python


print("Original Count:")
print(df.count())

# Remove duplicates
df = df.dropDuplicates()

print("Count After Removing Duplicates:")
print(df.count())
```

Handle Missing Values

```python

df = df.na.fill({"age": 0})

print("After Filling Null Values")
df.show()
```

Step 4: Filter Data

```python

print("Age > 25")
df.filter(col("age") > 25).show()
```

```python
print("Category = Electronics")
df.filter(col("category") == "Electronics").show()
```

```python
print("Region = North")
df.filter(col("region") == "North").show()
```

Step 5: Transform Data

```python

# Rename column
df = df.withColumnRenamed("sales", "total_sales")

# Change datatype
df = df.withColumn(
    "age",
    col("age").cast("integer")
)

print("After Transformation")
df.printSchema()
```

Step 6: Aggregation

```python

print("Total Rows:")
print(df.count())
```

```python
print("Average Sales")
df.select(avg("total_sales")).show()
```

```python
print("Minimum Sales")
df.select(min("total_sales")).show()
```

```python
print("Maximum Sales")
df.select(max("total_sales")).show()
```

Step 7: Group By Operations

```python
print("Sales by Region")

region_sales = df.groupBy("region") \
                 .agg(sum("total_sales").alias("total_sales"))

region_sales.show()
```

```python
print("Average Sales by Category")

df.groupBy("category") \
  .agg(avg("total_sales").alias("avg_sales")) \
  .show()
```

```python
print("Count by Category")

df.groupBy("category") \
  .count() \
  .show()
```

```python
import os
output_path = os.path.abspath("../output/results.csv")
region_sales_pd = region_sales.toPandas()
region_sales_pd.to_csv(output_path, index=False)

print("Results Saved Successfully")
```

## Wide Transformations and Shuffle

Wide Transformations:
Operations such as `groupBy()` and `join()` require data movement across partitions.

Shuffle:
The redistribution of data between partitions during wide transformations.

Impact:
- Increases network traffic
- Can slow Spark jobs
- Should be minimized for better performance
